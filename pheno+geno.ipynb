```python
import pandas as pd

# Load files
top10 = pd.read_csv("SCD_Top10_Complication_Details.csv")
demo = pd.read_csv("SCD_Demographics.csv")
geno = pd.read_csv("ALL_481_genotypes_wide.csv")

# Merge phenotype files
pheno = top10.merge(
    demo,
    on="person_id",
    how="left"
)

# Participant genotype columns
participant_cols = geno.columns[4:-1]

# Create unique variant names
variant_names = (
    geno["CHROM"].astype(str)
    + "_"
    + geno["POS"].astype(str)
    + "_"
    + geno["REF"]
    + "_"
    + geno["ALT"]
)

# Keep only participant genotype columns
geno2 = geno[participant_cols].copy()

# Set variant names as row index
geno2.index = variant_names

# Transpose to participant-level format
geno_t = geno2.T

geno_t.index.name = "person_id"

geno_t = geno_t.reset_index()

# Match ID type
geno_t["person_id"] = geno_t["person_id"].astype(int)

# Merge phenotype + genotype
analysis_df = pheno.merge(
    geno_t,
    on="person_id",
    how="inner"
)

# QC checks
print("Rows:", analysis_df.shape[0])
print("Columns:", analysis_df.shape[1])
print("Unique participants:", analysis_df["person_id"].nunique())

# Save
analysis_df.to_csv(
    "SCD_Genotype_Phenotype_Merged.csv",
    index=False
)

print("Saved successfully")
```

```python
import pandas as pd

df = pd.read_csv("SCD_Master_Analysis_Dataset.csv")

print(df.shape)
```

```python
for col in variant_cols:

    df[f"{col}_carrier"] = (
        df[col]
        .isin(["0|1", "1|0", "1|1"])
        .astype(int)
    )
```

```python
carrier_summary = []

for col in variant_cols:

    carrier_col = f"{col}_carrier"

    carrier_n = df[carrier_col].sum()

    noncarrier_n = len(df) - carrier_n

    carrier_pct = (
        carrier_n / len(df)
    ) * 100

    carrier_summary.append([
        col,
        carrier_n,
        noncarrier_n,
        round(carrier_pct, 2)
    ])

carrier_table = pd.DataFrame(
    carrier_summary,
    columns=[
        "Variant",
        "Carrier_N",
        "NonCarrier_N",
        "Carrier_Percent"
    ]
)

carrier_table = carrier_table.sort_values(
    "Carrier_N",
    ascending=False
)

print(carrier_table)
```

```python
carrier_table.sort_values("Carrier_N")
```

```python
complications = [
    "pulmonary_hypertension",
    "vitreous_haemorrhage",
    "liver_cirrhosis",
    "osteomyelitis",
    "aplastic_crisis",
    "preeclampsia",
    "sickle_retinopathy",
    "hand_foot_syndrome",
    "restrictive_lung_disease",
    "dactylitis"
]

comp_summary = []

for comp in complications:

    cases = df[comp].sum()

    controls = len(df) - cases

    pct = (cases / len(df)) * 100

    comp_summary.append([
        comp,
        cases,
        controls,
        round(pct, 2)
    ])

comp_table = pd.DataFrame(
    comp_summary,
    columns=[
        "Complication",
        "Cases",
        "Controls",
        "Percent"
    ]
)

comp_table = comp_table.sort_values(
    "Cases",
    ascending=False
)

print(comp_table)
```

```python
# Keep variants with at least 10 carriers

valid_variants = carrier_table.loc[
    carrier_table["Carrier_N"] >= 10,
    "Variant"
].tolist()

print("Number of variants retained:", len(valid_variants))
print(valid_variants)
```

```python
import pandas as pd
from scipy.stats import fisher_exact

results = []

for variant in valid_variants:

    carrier_col = f"{variant}_carrier"

    table = pd.crosstab(
        df[carrier_col],
        df["pulmonary_hypertension"]
    )

    # Require full 2x2 table
    if table.shape != (2, 2):
        continue

    odds_ratio, p_value = fisher_exact(table)

    results.append([
        variant,
        table.loc[0, 0],  # non-carriers without PH
        table.loc[0, 1],  # non-carriers with PH
        table.loc[1, 0],  # carriers without PH
        table.loc[1, 1],  # carriers with PH
        odds_ratio,
        p_value
    ])

ph_results = pd.DataFrame(
    results,
    columns=[
        "Variant",
        "NonCarrier_NoPH",
        "NonCarrier_PH",
        "Carrier_NoPH",
        "Carrier_PH",
        "Odds_Ratio",
        "P_Value"
    ]
)

ph_results = ph_results.sort_values("P_Value")

print(ph_results)

ph_results.to_csv(
    "Pulmonary_Hypertension_Fisher_Results.csv",
    index=False
)
```

```python
import numpy as np
import statsmodels.formula.api as smf

logistic_results = []

for variant in valid_variants:

    carrier_col = f"{variant}_carrier"

    try:

        model = smf.logit(
            f"pulmonary_hypertension ~ {carrier_col}",
            data=df
        ).fit(disp=0)

        beta = model.params[carrier_col]

        OR = np.exp(beta)

        CI_low = np.exp(
            model.conf_int().loc[carrier_col, 0]
        )

        CI_high = np.exp(
            model.conf_int().loc[carrier_col, 1]
        )

        p = model.pvalues[carrier_col]

        logistic_results.append([
            variant,
            OR,
            CI_low,
            CI_high,
            p
        ])

    except:
        pass

logistic_results = pd.DataFrame(
    logistic_results,
    columns=[
        "Variant",
        "OR",
        "CI_Lower",
        "CI_Upper",
        "P_Value"
    ]
)

logistic_results = logistic_results.sort_values(
    "P_Value"
)

print(logistic_results)

logistic_results.to_csv(
    "Pulmonary_Hypertension_Logistic_Results.csv",
    index=False
)
```

```python
from statsmodels.stats.multitest import multipletests

logistic_results["FDR_P"] = multipletests(
    logistic_results["P_Value"],
    method="fdr_bh"
)[1]

logistic_results = logistic_results.sort_values(
    "FDR_P"
)
print(logistic_results)
```

```python
from scipy.stats import fisher_exact
import pandas as pd

complications = [
    "vitreous_haemorrhage",
    "liver_cirrhosis",
    "osteomyelitis",
    "aplastic_crisis",
    "preeclampsia"
]

all_results = []

for comp in complications:

    for variant in valid_variants:

        carrier_col = f"{variant}_carrier"

        table = pd.crosstab(
            df[carrier_col],
            df[comp]
        )

        if table.shape != (2, 2):
            continue

        OR, p = fisher_exact(table)

        all_results.append([
            comp,
            variant,
            table.loc[0,0],
            table.loc[0,1],
            table.loc[1,0],
            table.loc[1,1],
            OR,
            p
        ])

results_df = pd.DataFrame(
    all_results,
    columns=[
        "Complication",
        "Variant",
        "NonCarrier_NoComp",
        "NonCarrier_Comp",
        "Carrier_NoComp",
        "Carrier_Comp",
        "Odds_Ratio",
        "P_Value"
    ]
)

results_df = results_df.sort_values("P_Value")

print(results_df.head(20))

results_df.to_csv(
    "All_Complication_Fisher_Results.csv",
    index=False
)
```

```python
import pandas as pd
import statsmodels.formula.api as smf

results = []

for trait in lab_traits:

    for var in valid_variants:

        carrier_col = f"{var}_carrier"

        if carrier_col not in df.columns:
            continue

        temp = df[[trait, carrier_col]].dropna()

        if len(temp) < 30:
            continue

        try:

            model = smf.ols(
                f'Q("{trait}") ~ {carrier_col}',
                data=temp
            ).fit()

            beta = model.params[carrier_col]
            p = model.pvalues[carrier_col]

            ci = model.conf_int().loc[carrier_col]

            results.append([
                trait,
                var,
                len(temp),
                beta,
                ci[0],
                ci[1],
                p
            ])

        except Exception:
            continue

carrier_lab_results = pd.DataFrame(
    results,
    columns=[
        "Trait",
        "Variant",
        "N",
        "Beta_carrier_effect",
        "CI_Lower",
        "CI_Upper",
        "P_Value"
    ]
)

carrier_lab_results = carrier_lab_results.sort_values("P_Value")

print(carrier_lab_results.head(20))
```

```python
import pandas as pd

master = pd.read_csv("SCD_Master_Analysis_Dataset.csv")

print(master.shape)
```

```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf

labs = [
    "Fetal hemoglobin",
    "Hemoglobin",
    "Creatinine",
    "WBC (Leukocytes)",
    "Neutrophils",
    "Monocytes",
    "Basophils", 
    "Diastolic blood pressure",
    "Systolic blood pressure"
]

complications = [
    "pulmonary_hypertension",
    "vitreous_haemorrhage",
    "liver_cirrhosis",
    "osteomyelitis",
    "aplastic_crisis",
    "preeclampsia"
]

results = []

for lab in labs:

    for comp in complications:

        df = master[[lab, comp]].dropna()

        if len(df) < 20:
            continue

        if df[comp].nunique() < 2:
            continue

        try:

            model = smf.logit(
                f"{comp} ~ Q('{lab}')",
                data=df
            ).fit(disp=0)

            beta = model.params.iloc[1]
            p = model.pvalues.iloc[1]

            OR = np.exp(beta)

            results.append([
                lab,
                comp,
                len(df),
                beta,
                OR,
                p
            ])

        except:
            pass

lab_comp_results = pd.DataFrame(
    results,
    columns=[
        "Lab",
        "Complication",
        "N",
        "Beta",
        "Odds_Ratio",
        "P_Value"
    ]
)

lab_comp_results.sort_values(
    "P_Value"
).head(20)
```

```python
from statsmodels.stats.multitest import multipletests

lab_comp_results["FDR_P"] = multipletests(
    lab_comp_results["P_Value"],
    method="fdr_bh"
)[1]

lab_comp_results.sort_values(
    "FDR_P"
).head(20)
```

```python
import pandas as pd

master = pd.read_csv(
    "SCD_Master_Analysis_Dataset.csv"
)

print(master.shape)
master.head()
```

```python
variant_cols = [
    "VCAM_chr1_100718269_T_C",
    "HBBcluster_chr11_5239228_T_C",
    "HBBcluster_chr11_5242453_C_T",
    "HBBcluster_chr11_5248569_A_C",
    "HBBcluster_chr11_5248569_A_T",
    "HBG2_chr11_5254939_G_A",
    "APOE_chr19_44908684_T_C",
    "APOE_chr19_44908822_C_A",
    "APOE_chr19_44908822_C_T",
    "BCL11A_chr2_60487726_A_C",
    "BCL11A_chr2_60490908_T_G",
    "BCL11A_chr2_60498316_G_A",
    "BCL11A_chr2_60498316_G_C",
    "APOL1_chr22_36265860_A_G",
    "APOL1_chr22_36265988_T_G",
    "APOL1_chr22_36265996_ATAATT_A",
    "APOL1_chr22_36265996_ATAATT_GTAATT",
    "TNF_chr6_31575254_G_A",
    "HBS1L_MYB_chr6_135097526_GC_CC",
    "HBS1L_MYB_chr6_135097526_GC_G",
    "HBS1L_MYB_chr6_135097880_T_C"
]

for var in variant_cols:

    master[var + "_carrier"] = (
        master[var]
        .astype(str)
        .isin(["0|1", "1|0", "1|1"])
        .astype(int)
    )
```

```python
from scipy.stats import fisher_exact
import pandas as pd
import numpy as np

complications = [
    "osteomyelitis",
    "aplastic_crisis",
    "vitreous_haemorrhage",
    "liver_cirrhosis",
    "preeclampsia"
]

carrier_cols = [
    c for c in master.columns
    if c.endswith("_carrier")
]

results = []

for variant in carrier_cols:

    for comp in complications:

        df = master[[variant, comp]].dropna()

        table = pd.crosstab(
            df[variant],
            df[comp]
        )

        if table.shape != (2,2):
            continue

        OR, p = fisher_exact(table)

        results.append([
            comp,
            variant.replace("_carrier",""),
            table.loc[0,0],
            table.loc[0,1],
            table.loc[1,0],
            table.loc[1,1],
            OR,
            p
        ])

all_fisher = pd.DataFrame(
    results,
    columns=[
        "Complication",
        "Variant",
        "NonCarrier_NoComp",
        "NonCarrier_Comp",
        "Carrier_NoComp",
        "Carrier_Comp",
        "Odds_Ratio",
        "P_Value"
    ]
)

all_fisher.to_csv(
    "All_Complication_Fisher_Results.csv",
    index=False
)
```
